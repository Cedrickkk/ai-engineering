### Chat Completions API

It is the simplest way to call an LLM and was invented by OpenAI

```python
from openai import OpenAI
from openai.types.chat import ChatCompletion

ollama = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama",
)
```

```python
response: ChatCompletion = ollama.chat.completions.create(
    model="llama3.2",
    messages=[
        {
            "role": "system",
            "content": "Tell me a fun fact."
        }
    ]
)

print(response.model_dump_json(indent=4))
```

### Making an http call to OpenAPI endpoint with `requests`

```python
import requests

from typing import Any

headers = {
    "Authorization": f"Bearer __some__api__key__here__",
    "Content-Type": "application/json",
}

payload: dict[str, Any] = {
    "model": "openai-model-here",
    "messages": [
        {
            "role": "system",
            "content": "Tell me a fun fact."
        }
    ]
}

http_call_response = requests.post(
    url="https://api.openai.com/v1/chat/completions",
    headers=headers,
    json=payload
)

http_call_response.json()["choices"][0]["messages"]["content"]
```

### `openai` package

It is a Python Client Library that is a wrapper around making a call to an http endpoint shown above.

It allows us to work in Python instead of messing around with messy json objects.

```python
# Creating an OpenAI client

openai = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama",
)

response: ChatCompletion = openai.chat.completions.create(
    model="llama3.2",
    messages=[
        {
            "role": "system",
            "content": "Tell me something mind blowing."
        }
    ]
)

print(response.choices[0].message.content)
```
