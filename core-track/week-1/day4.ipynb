# The Illusion of Memory

**LLMs are stateless**: In standard Large Language Models (LLMs) like ChatGPT, Claude, or Llama, no persistent learning happens during live inference (chatting).

The model itself has no memory between messages. Every time you send a message, the app resends the entire conversation history (all previous messages + the new one) as one input, and the model re-reads it all fresh each time.

### Setup OpenAI

```python
from openai import OpenAI
from openai.types.chat import ChatCompletion, ChatCompletionMessageParam

ollama = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)
```

### Construct a message to Ollama

```python

messages: list[ChatCompletionMessageParam] = [
    {
        "role": "system",
        "content": "You are a helpful assistant",
    },
    {
        "role": "user",
        "content": "Hi! I'm Cedrick!"
    }
]
```

### Create a chat and print the response

```python
response: ChatCompletion = ollama.chat.completions.create(
    model="llama3.2",
    messages=messages
)

print(response.choices[0].message.content)
```

### Follow-up question

```python
messages: list[ChatCompletionMessageParam] = [
    {
        "role": "system",
        "content": "You are a helpful assistant",
    },
    {
        "role": "user",
        "content": "What is my name?"
    }
]  
```

### Construct another conversation

```python
response: ChatCompletion = ollama.chat.completions.create(
    model="llama3.2",
    messages=messages
)
```

### Stateless response

```python
print(response.choices[0].message.content)
```

### In summary

- Every call to an LLM is stateless
- We pass in the entire converstation so far in the input, every time
- This gives the illusion in the LLM has memory
- But this is a trick; it's a by-product of providing the entire conversation, every time
- An LLM just predicts the most likely next tokens in the sequence
