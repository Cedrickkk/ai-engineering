### Quick call to local Frontier model

### Types of Prompts

- **System Prompt**: tells the model what task they are performing and what tone they should use
- **User Prompt**: the conversation starter that the model should reply to

**System Prompt**

```python
system_prompt = """
You are a humorous AI Engineer and a certified Gen-Z that analyzes the contents of a website.
Provide a short humorous summary, ignoring text that might have navigate to links to other site.
Respond only in markdown, do not wrap the markdown in a code-block.
"""
```

**User Prompt**

```python
user_prompt = """
Here are the contents of a website.
Provide a short summary of this website.
"""
```

```python
test_messages: list[ChatCompletionMessageParam] = [
    {
        "role": "system",
        "content": "You are a helpful assistant",
    },
    {
        "role": "user",
        "content": "What is 2 + 2?"
    }
]

test_response = openai.chat.completions.create(
    model="llama3.2",
    messages=test_messages
)

print(test_response.choices[0].message.content)
```

```python
def message_for(website: str) -> list[ChatCompletionMessageParam]:
    return [
        {
            "role": "system",
            "content": system_prompt
        }, 
        {
            "role": "user",
            "content": user_prompt + website
        }
    ]
```

```python
message_for(ed)
```

### Bringing it all together!

```python
def summarize(url: str) -> str | None:
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model="llama3.2",
        messages=message_for(website)
    )
    return response.choices[0].message.content
```

```python
summarize("https://edwarddonner.com")
```

```python
def display_summary(url: str) -> None:
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://edwarddonner.com")
```
